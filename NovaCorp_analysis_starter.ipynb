```python
# STEP 1: Funnel & driver split
emp = datasets["employees"].copy()
attr = datasets["attrition_log"].copy()
emp["hire_date"] = pd.to_datetime(emp["hire_date"], errors="coerce")
emp["exit_date"] = pd.to_datetime(emp["exit_date"], errors="coerce")
attr["exit_date"] = pd.to_datetime(attr["exit_date"], errors="coerce")

ref_date = emp["exit_date"].max()
headcount = (emp["hire_date"] <= ref_date).sum()
n_all = len(attr)
n_vol = (attr["exit_type"] == "voluntary").sum()
n_reg = (attr["regrettable_flag"] == True).sum()

funnel = pd.DataFrame({
    "stage": ["Headcount FY25", "All exits", "Voluntary", "Regrettable"],
    "count": [headcount, n_all, n_vol, n_reg],
    "rate_vs_headcount_pct": [
        100.0,
        round(n_all/headcount*100, 2),
        round(n_vol/headcount*100, 2),
        round(n_reg/headcount*100, 2),
    ],
})
display(funnel)

print("
Push vs Pull split (%):")
print(attr["pathway"].value_counts(normalize=True).mul(100).round(2))

print("
Top exit reasons — all (%):")
print(attr["stated_exit_reason"].value_counts(normalize=True).mul(100).round(2).head(5))

print("
Top exit reasons — regrettable (%):")
print(attr.loc[attr["regrettable_flag"], "stated_exit_reason"].value_counts(normalize=True).mul(100).round(2).head(5))
```

```python
# STEP 2: Department & role-level hotspots
reg_full = attr.loc[attr["regrettable_flag"]].merge(
    emp[["employee_id", "department", "role_level"]],
    on="employee_id", how="left",
)

dept_hc = emp.groupby("department").size()
dept_reg = reg_full.groupby("department").size()
dept_tbl = pd.DataFrame({
    "headcount": dept_hc,
    "regrettable_exits": dept_reg.fillna(0).astype(int),
}).assign(
    reg_rate_pct=lambda d: round(d["regrettable_exits"]/d["headcount"]*100, 2),
    share_of_reg_pct=lambda d: round(d["regrettable_exits"]/n_reg*100, 2),
).sort_values("regrettable_exits", ascending=False)
display(dept_tbl)

lvl_hc = emp.groupby("role_level").size()
lvl_reg = reg_full.groupby("role_level").size()
lvl_tbl = pd.DataFrame({
    "headcount": lvl_hc,
    "regrettable_exits": lvl_reg.fillna(0).astype(int),
}).assign(
    reg_rate_pct=lambda d: round(d["regrettable_exits"]/d["headcount"]*100, 2),
    share_of_reg_pct=lambda d: round(d["regrettable_exits"]/n_reg*100, 2),
).sort_index()
display(lvl_tbl)

frontline = lvl_tbl.loc[lvl_tbl.index.isin([1,2,3]), "regrettable_exits"].sum()
print(f"
Frontline (L1-L3) share of regrettable: {frontline/n_reg*100:.2f}%")
```

```python
# STEP 3: Statistical tests — z-test per dept, Welch's t for engagement, chi-square for HiPO
from scipy import stats

# 3a) Per-dept z-test vs 9% benchmark
rows = []
for dept, row in dept_tbl.iterrows():
    n = int(row["headcount"])
    x = int(row["regrettable_exits"])
    p_hat = x/n
    p0 = 0.09
    se = (p0*(1-p0)/n)**0.5
    z = (p_hat - p0)/se
    p = 1 - stats.norm.cdf(z)
    rows.append({
        "department": dept, "n": n, "reg_exits": x,
        "rate_pct": round(p_hat*100, 2),
        "z": round(z, 3), "p_value": round(p, 4),
        "sig_above_9pct": p < 0.05,
    })
z_df = pd.DataFrame(rows).sort_values("rate_pct", ascending=False)
display(z_df)

# 3b) Engagement: regrettable vs retained (last wave)
eng = datasets["engagement"].copy()
score_cols = [
    "manager_effectiveness","psychological_safety","recognition",
    "career_development","senior_leadership_trust","purpose_meaning",
    "wellbeing","confidence_in_role_future",
]
last_wave = (
    eng.sort_values(["employee_id","wave_number"])
       .groupby("employee_id").tail(1)[["employee_id"]+score_cols]
       .merge(attr[["employee_id","regrettable_flag"]], on="employee_id", how="left")
)
last_wave["regrettable"] = last_wave["regrettable_flag"].fillna(False)

t_rows = []
for c in score_cols:
    g = last_wave.loc[last_wave["regrettable"], c].dropna()
    r = last_wave.loc[~last_wave["regrettable"], c].dropna()
    t, p = stats.ttest_ind(g, r, equal_var=False)
    pooled = pd.concat([g, r]).std()
    d = (g.mean() - r.mean()) / pooled if pooled > 0 else 0
    t_rows.append({
        "dimension": c,
        "reg_mean": round(g.mean(), 3),
        "ret_mean": round(r.mean(), 3),
        "diff": round(g.mean()-r.mean(), 3),
        "t": round(t, 3), "p_value": round(p, 6),
        "cohens_d": round(d, 3),
        "n_reg": len(g), "n_ret": len(r),
    })
t_df = pd.DataFrame(t_rows).sort_values("p_value")
display(t_df)

# 3c) HiPO / promotion chi-square
for flag in ["hipo_flag", "promotion_eligible"]:
    base = emp[flag].mean()
    reg_rate = reg_full.merge(emp[["employee_id", flag]], on="employee_id")[flag].mean()
    x = int(reg_full.merge(emp[["employee_id", flag]], on="employee_id")[flag].sum())
    n = len(reg_full)
    exp = base * n
    chi2 = (x - exp)**2 / exp if exp > 0 else 0
    p = 1 - stats.chi2.cdf(chi2, df=1)
    lift = (reg_rate/base - 1)*100 if base > 0 else 0
    print(f"{flag}: baseline={base*100:.2f}%, regrettable={reg_rate*100:.2f}%, "
          f"lift=+{lift:.1f}%, chi2={chi2:.2f}, p={p:.4f}")
```

```python
# STEP 4: Cost by department
total_cost = 45_800_000
reg_share = n_reg / n_all
reg_cost = total_cost * reg_share
print(f"Total annual attrition cost: A${total_cost:,}")
print(f"Regrettable share: {reg_share*100:.2f}%")
print(f"Estimated regrettable cost: A${reg_cost:,.0f}")

dept_cost = dept_tbl.assign(
    est_cost_aud=lambda d: round(d["share_of_reg_pct"]/100 * reg_cost, 0)
).sort_values("est_cost_aud", ascending=False)
display(dept_cost[["regrettable_exits", "share_of_reg_pct", "reg_rate_pct", "est_cost_aud"]])
```

```python
# LENS 1: Timing & Tenure
# When do exits cluster? What tenure bands attrite most?
import numpy as np

attr_full = attr.merge(emp[["employee_id","hire_date","department","role_level"]], on="employee_id", how="left")
attr_full["tenure_at_exit_months"] = ((attr_full["exit_date"] - attr_full["hire_date"]).dt.days / 30.25).round(1)

# Tenure bands
def tenure_band(m):
    if m < 12: return "<1 yr"
    if m < 24: return "1-2 yrs"
    if m < 60: return "2-5 yrs"
    if m < 120: return "5-10 yrs"
    return "10+ yrs"
attr_full["tenure_band"] = attr_full["tenure_at_exit_months"].apply(tenure_band)

ten = attr_full["tenure_band"].value_counts(normalize=True).mul(100).round(2)
print("Tenure at exit — all (%):")
print(ten)

reg_ten = attr_full.loc[attr_full["regrettable_flag"], "tenure_band"].value_counts(normalize=True).mul(100).round(2)
print("\nTenure at exit — regrettable (%):")
print(reg_ten)

# Exit by quarter
attr_full["exit_quarter"] = attr_full["exit_date"].dt.to_period("Q").astype(str)
q = attr_full["exit_quarter"].value_counts().sort_index()
print("\nExits by quarter:")
print(q)

# Regrettable rate by tenure band vs overall
ten_reg = attr_full.groupby("tenure_band").agg(
    exits=("employee_id","count"),
    reg=("regrettable_flag","sum"),
).assign(reg_rate=lambda d: round(d["reg"]/d["exits"]*100,2))
print("\nRegrettable rate by tenure band:")
display(ten_reg)
```

```python
# LENS 2: Manager as unit of analysis
# Which managers lose the most regrettable talent?
mgr = attr_full.loc[attr_full["regrettable_flag"]].groupby("manager_id_at_exit").agg(
    regrettable_exits=("employee_id","count"),
    depts=("department", lambda x: x.nunique()),
).sort_values("regrettable_exits", ascending=False)

print("Top 15 managers by regrettable exits:")
print(mgr.head(15).to_string())

multi = (mgr["regrettable_exits"] >= 2).sum()
print(f"\nManagers with 2+ regrettable exits: {multi} ({multi/len(mgr)*100:.1f}% of exit-managers)")

from scipy import stats as st
n_mgr = len(mgr)
total_reg = mgr["regrettable_exits"].sum()
expected = total_reg / n_mgr
observed_top10 = mgr.head(10)["regrettable_exits"].sum()
chi2_conc = (observed_top10 - 10*expected)**2 / (10*expected) + \
            ((total_reg - observed_top10) - (n_mgr-10)*expected)**2 / ((n_mgr-10)*expected)
p_conc = 1 - st.chi2.cdf(chi2_conc, df=1)
print(f"\nTop-10 managers hold {observed_top10/total_reg*100:.1f}% of regrettable exits")
print(f"Chi-square concentration test: chi2={chi2_conc:.2f}, p={p_conc:.6f}")
```

```python
# LENS 3: Career pathway — acting appointments, promotion eligibility
emp_c = emp.copy()
emp_c["is_leaver"] = emp_c["exit_date"].notna()

acting_rate = emp_c.groupby("acting_appointment")["is_leaver"].mean().mul(100).round(2)
print("Attrition rate by acting_appointment (%):")
print(acting_rate)

ct = pd.crosstab(emp_c["acting_appointment"], emp_c["is_leaver"])
chi2_a, p_a, _, _ = st.chi2_contingency(ct)
print(f"\nActing-appointment chi2={chi2_a:.2f}, p={p_a:.6f}")

ct2 = pd.crosstab(emp_c["promotion_eligible"], emp_c["is_leaver"])
chi2_p, p_p, _, _ = st.chi2_contingency(ct2)
print(f"Promotion-eligible chi2={chi2_p:.2f}, p={p_p:.6f}")

promo_leavers = emp_c.loc[emp_c["is_leaver"], "promotion_eligible"].mean()*100
promo_stayers = emp_c.loc[~emp_c["is_leaver"], "promotion_eligible"].mean()*100
print(f"\nPromotion-eligible: leavers={promo_leavers:.2f}%, stayers={promo_stayers:.2f}%")
```

```python
# LENS 4: Compensation — compa-ratio, salary at exit
emp_c2 = emp.copy()
emp_c2["is_leaver"] = emp_c2["exit_date"].notna()

emp_c2["comp_band"] = pd.cut(emp_c2["compa_ratio"], bins=[0, 0.85, 0.95, 1.05, 1.15, 10], labels=["<0.85","0.85-0.95","0.95-1.05","1.05-1.15","1.15+"])
comp_attr = emp_c2.groupby("comp_band", observed=True)["is_leaver"].mean().mul(100).round(2)
print("Attrition rate by compa-ratio band (%):")
print(comp_attr)

t_c, p_c = st.ttest_ind(
    emp_c2.loc[emp_c2["is_leaver"], "compa_ratio"].dropna(),
    emp_c2.loc[~emp_c2["is_leaver"], "compa_ratio"].dropna(),
    equal_var=False,
)
print(f"\nCompa-ratio t-test: t={t_c:.3f}, p={p_c:.6f}")
print(f"Mean compa-ratio leavers: {emp_c2.loc[emp_c2['is_leaver'],'compa_ratio'].mean():.3f}")
print(f"Mean compa-ratio stayers: {emp_c2.loc[~emp_c2['is_leaver'],'compa_ratio'].mean():.3f}")
```

```python
# LENS 5: Legacy entity / acquisition cohorts
emp_c3 = emp.copy()
emp_c3["is_leaver"] = emp_c3["exit_date"].notna()

leg = emp_c3.groupby("legacy_entity_code").agg(
    headcount=("employee_id","count"),
    leavers=("is_leaver","sum"),
).assign(attr_rate=lambda d: round(d["leavers"]/d["headcount"]*100, 2))
print("Attrition by legacy entity (%):")
display(leg.sort_values("attr_rate", ascending=False))

ct3 = pd.crosstab(emp_c3["legacy_entity_code"], emp_c3["is_leaver"])
chi2_l, p_l, _, _ = st.chi2_contingency(ct3)
print(f"Legacy-entity chi2={chi2_l:.2f}, p={p_l:.6f}")

print(f"\ndays_to_fill range: {emp['days_to_fill'].min()}-{emp['days_to_fill'].max()}")
print(f"Mean days_to_fill leavers: {emp_c3.loc[emp_c3['is_leaver'],'days_to_fill'].mean():.1f}")
print(f"Mean days_to_fill stayers: {emp_c3.loc[~emp_c3['is_leaver'],'days_to_fill'].mean():.1f}")
t_d, p_d = st.ttest_ind(
    emp_c3.loc[emp_c3["is_leaver"], "days_to_fill"].dropna(),
    emp_c3.loc[~emp_c3["is_leaver"], "days_to_fill"].dropna(),
    equal_var=False,
)
print(f"days_to_fill t={t_d:.3f}, p={p_d:.6f}")
```

```python
# LENS 6: Performance trajectory — last review before exit
perf["review_date"] = pd.to_datetime(perf["review_date"], errors="coerce")
attr_perf = attr.merge(perf, on="employee_id", how="left")

last_review = attr_perf.sort_values(["employee_id","review_date"]).groupby("employee_id").tail(1)
last_review = last_review.loc[last_review["employee_id"].isin(attr["employee_id"])]

print("Last performance rating of leavers:")
print(last_review["performance_rating"].value_counts(normalize=True).mul(100).round(2))

emp_perf = emp.merge(perf, on="employee_id", how="left")
emp_last = emp_perf.sort_values(["employee_id","review_date"]).groupby("employee_id").tail(1)
print("\nLatest performance rating — all employees:")
print(emp_last["performance_rating"].value_counts(normalize=True).mul(100).round(2))

leaver_scores = perf[perf["employee_id"].isin(attr.loc[attr["regrettable_flag"],"employee_id"])].sort_values(["employee_id","review_date"])
trend = leaver_scores.groupby("employee_id")["goal_achievement_score"].agg(["first","last"]).dropna()
trend["delta"] = trend["last"] - trend["first"]
print(f"\nGoal score trend for regrettable leavers:")
print(f"  Mean first score: {trend['first'].mean():.1f}")
print(f"  Mean last score: {trend['last'].mean():.1f}")
print(f"  Mean delta: {trend['delta'].mean():.2f}")
t_g, p_g = st.ttest_rel(trend["first"], trend["last"])
print(f"  Paired t-test: t={t_g:.3f}, p={p_g:.6f}")
```

```python
# LENS 7: Survey wave momentum — did scores DROP before exit?
eng_sorted = eng.sort_values(["employee_id","wave_number"])

leaver_ids = attr.loc[attr["regrettable_flag"], "employee_id"]
leaver_eng = eng_sorted[eng_sorted["employee_id"].isin(leaver_ids)]

last_two = leaver_eng.groupby("employee_id").tail(2)
last_two["wave_pos"] = last_two.groupby("employee_id").cumcount(ascending=False)
pivoted = last_two.pivot_table(index="employee_id", columns="wave_pos", values=score_cols)

momentum = pd.DataFrame()
for c in score_cols:
    if c in pivoted.columns.get_level_values(0):
        try:
            newest = pivoted[(c, 0)]
            prev = pivoted[(c, 1)]
            delta = newest - prev
            momentum[c] = delta
        except KeyError:
            pass

print("Mean score change (last wave - previous wave) for regrettable leavers:")
print(momentum.mean().round(3))

print("\nOne-sample t-test (H0: delta = 0):")
for c in momentum.columns:
    vals = momentum[c].dropna()
    t, p = st.ttest_1samp(vals, 0)
    print(f"  {c:30s}: mean={vals.mean():+.3f}, t={t:+.3f}, p={p:.4f}, n={len(vals)}")
```

```python
# LENS 8: Hiring source quality
emp_c4 = emp.copy()
emp_c4["is_leaver"] = emp_c4["exit_date"].notna()

src = emp_c4.groupby("hire_source").agg(
    headcount=("employee_id","count"),
    leavers=("is_leaver","sum"),
).assign(attr_rate=lambda d: round(d["leavers"]/d["headcount"]*100, 2))
print("Attrition by hire_source (%):")
display(src.sort_values("attr_rate", ascending=False))

ct5 = pd.crosstab(emp_c4["hire_source"], emp_c4["is_leaver"])
chi2_s, p_s, _, _ = st.chi2_contingency(ct5)
print(f"Hire-source chi2={chi2_s:.2f}, p={p_s:.6f}")
```

```python
# LENS 9: Diversity & inclusion — gender, age_band, cultural_background
emp_c5 = emp.copy()
emp_c5["is_leaver"] = emp_c5["exit_date"].notna()

for col in ["gender", "age_band", "cultural_background"]:
    g = emp_c5.groupby(col).agg(
        headcount=("employee_id","count"),
        leavers=("is_leaver","sum"),
    ).assign(attr_rate=lambda d: round(d["leavers"]/d["headcount"]*100, 2))
    print(f"\nAttrition by {col}:")
    display(g.sort_values("attr_rate", ascending=False))
    ct = pd.crosstab(emp_c5[col], emp_c5["is_leaver"])
    chi2, p, _, _ = st.chi2_contingency(ct)
    print(f"{col} chi2={chi2:.2f}, p={p:.6f}")
```

```python
# LENS 10: Predictive risk score for active employees
emp_active = emp[emp["status"]=="active"].copy() if "status" in emp.columns else emp[emp["exit_date"].isna()].copy()

last_eng = eng.sort_values(["employee_id","wave_number"]).groupby("employee_id").tail(1)
emp_active = emp_active.merge(last_eng[["employee_id"]+score_cols], on="employee_id", how="left")

emp_active["low_engagement"] = emp_active[score_cols].mean(axis=1) < 3.5
emp_active["is_hipo"] = emp_active["hipo_flag"].fillna(False)
emp_active["underpaid"] = emp_active["compa_ratio"] < 0.9
emp_active["is_frontline"] = emp_active["role_level"].isin([1,2,3])

emp_active["risk_score"] = (
    emp_active["low_engagement"].astype(int)
    + emp_active["is_hipo"].astype(int)
    + emp_active["underpaid"].astype(int)
    + emp_active["is_frontline"].astype(int)
)

print("Risk score distribution (active employees):")
print(emp_active["risk_score"].value_counts().sort_index())

high_risk = (emp_active["risk_score"] >= 3).sum()
print(f"\nHigh-risk (score >= 3) active employees: {high_risk} ({high_risk/len(emp_active)*100:.1f}%)")

known_leavers = emp[emp["exit_date"].notna()].copy()
last_eng_l = eng.sort_values(["employee_id","wave_number"]).groupby("employee_id").tail(1)
known_leavers = known_leavers.merge(last_eng_l[["employee_id"]+score_cols], on="employee_id", how="left")
known_leavers["low_engagement"] = known_leavers[score_cols].mean(axis=1) < 3.5
known_leavers["is_hipo"] = known_leavers["hipo_flag"].fillna(False)
known_leavers["underpaid"] = known_leavers["compa_ratio"] < 0.9
known_leavers["is_frontline"] = known_leavers["role_level"].isin([1,2,3])
known_leavers["risk_score"] = (
    known_leavers["low_engagement"].astype(int)
    + known_leavers["is_hipo"].astype(int)
    + known_leavers["underpaid"].astype(int)
    + known_leavers["is_frontline"].astype(int)
)

print("\nRisk score — known leavers:")
print(known_leavers["risk_score"].value_counts(normalize=True).mul(100).round(2).sort_index())
print("\nRisk score — active:")
print(emp_active["risk_score"].value_counts(normalize=True).mul(100).round(2).sort_index())

from scipy.stats import mannwhitneyu
u, p_mw = mannwhitneyu(known_leavers["risk_score"], emp_active["risk_score"], alternative="greater")
print(f"\nMann-Whitney U (leavers > active): U={u:.0f}, p={p_mw:.6f}")
print(f"Mean risk score leavers: {known_leavers['risk_score'].mean():.2f}")
print(f"Mean risk score active: {emp_active['risk_score'].mean():.2f}")
```
